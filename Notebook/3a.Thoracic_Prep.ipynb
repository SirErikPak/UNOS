### Plot

```python
# get table into DataFrame
data = uf.category_contingency(df, 'TransplantYear', 'RecipientStatus')
# exclude the last row
data = data.iloc[:-1]

# 'Dead %' or 'Living %'
data = data[['Dead %', 'Living %']]

# Plotting the bar plot
ax = data.plot(kind='bar', figsize=(12, 6))
ax.set_title('Dead % and Living % vs Transplant Year')
ax.set_xlabel('TransplantYear')
ax.set_ylabel('Percentage')
ax.set_xticks(range(len(data.index)))
ax.set_xticklabels(data.index, rotation=0)
ax.grid(True)

# ── save to disk ─────────────────────────────────────────────
plt.tight_layout()                     # trims excess white space
plt.savefig('../Image/dead_living_vs_transplant_year.png',  # file name & format
            dpi=600,                   # high‑resolution
            bbox_inches='tight')       # crops to the actual plot area

# Optional: show it interactively
plt.show()
```
