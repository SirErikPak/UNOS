```python
import pandas as pd
import numpy as np
from sklearn.experimental import enable_iterative_imputer  # noqa
from sklearn.impute import IterativeImputer
from sklearn.linear_model import LinearRegression
from sklearn.linear_model import BayesianRidge


# Example toy data (3 features, some NaNs)
df = pd.DataFrame({
    "HR":  [70, 85, np.nan, 90, 76],
    "MAP": [93, np.nan, 100, 105, 88],
    "CI":  [2.6, 3.0, 3.2, np.nan, 2.8],
})
```

```python
imputer = IterativeImputer(
    estimator=LinearRegression(),
    max_iter=10,
    initial_strategy="mean",
    random_state=42,
)
```

```python
imputer = IterativeImputer(
    estimator=LinearRegression(),
    max_iter=10,
    initial_strategy="median",
    random_state=42,
)
```

```python
imputed_array = imputer.fit_transform(df)
df_imputed = pd.DataFrame(imputed_array, columns=df.columns)

print(df_imputed)
```

```python
from sklearn.metrics import mean_squared_error

mask_not_nan = ~df.isna()          # True wherever original value is present
orig_vals    = df[mask_not_nan]    # drop NaNs


for n in [3, 5, 10, 20]:
    imputer = IterativeImputer(estimator=BayesianRidge(), n_nearest_features=n, random_state=0)
    df_imputed = imputer.fit_transform(df)
    mse = mean_squared_error(orig_vals.values, df_imputed)
    print(f"n_nearest_features={n} => MSE: {mse:.4f}")
```
